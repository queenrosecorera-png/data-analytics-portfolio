```python
import matplotlib.pyplot as plt

sentiment_counts = df_clean["sentiment"].value_counts()
sentiment_percent = sentiment_counts / sentiment_counts.sum() * 100

ax = sentiment_counts.plot(kind="bar")

for i, (count, percent) in enumerate(zip(sentiment_counts, sentiment_percent)):
    ax.text(i, count, f"{percent:.1f}%", ha="center", va="bottom")

plt.title("Overall Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Posts")
plt.xticks(rotation=0)
plt.show()
```
