```python
region = (
    df.groupby(["City"])["InvestmentAmount_USD"]
      .sum()
      .sort_values(ascending=False)
)

region
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
```

```python
date = pd.to_datetime(df["Date"],errors="coerce")
```

```python
df.info()
```

```python
df["Year"] = df["Date"].dt.year
```

```python
df.head()
```

```python
industry_funding = (
    df.groupby("Industry")["InvestmentAmount_USD"]
      .sum()
      .sort_values(ascending=False)
)

industry_funding
```

```python
investment =industry_funding.sort_values(ascending=False).plot(kind="bar")

plt.title("Total Startup Investment by Industry")
plt.xlabel("Industry")
plt.ylabel("Total Investment (USD)")

plt.show()
```

```python
investment_type_funding = (
    df.groupby("InvestmentType")["InvestmentAmount_USD"]
      .sum()
      .sort_values(ascending=False)
)
investment_type_funding
```

```python
yearly_funding = (
    df.groupby("Year")["InvestmentAmount_USD"]
      .sum()
      .sort_index()
)
yearly_funding
```

```python
plt.figure(figsize=(14,8))

plt.subplot(2,2,1)
region.plot(kind="bar",color="red")
plt.title("City")
plt.ylabel("Investment (USD)")

plt.subplot(2,2,2)
industry_funding.plot(kind="bar",color="blue")
plt.title("Industry Funding")
plt.ylabel("Investment (USD)")

plt.subplot(2,2,3)
investment_type_funding.plot(kind="bar",color="blue")
plt.title("Investment by Funding Stage")
plt.ylabel("Investment (USD)")

plt.subplot(2,2,4)
yearly_funding.plot(kind="line", marker="o")
plt.title("Startup Investment Growth Over Time")
plt.ylabel("Investment (USD)")



plt.tight_layout()
plt.show()
```

```python
investor_count = df['Investors'].value_counts()
investor_count
```

```python
investor_count.head(10).plot(kind="barh")

plt.title("Top 10 Investors by Number of Investments")
plt.xlabel("Investor")
plt.ylabel("Number of Investments")
#plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
investor_funding = df.groupby("Investors")["InvestmentAmount_USD"].sum()
investor_funding
```

```python
top_startups = (
    df.groupby("Startup")["InvestmentAmount_USD"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_startups
```

```python
investor_funding.head(10).sort_values().plot(kind="barh")

plt.title("Top 10 Investors by Total Investment")
plt.xlabel("Total Investment")
plt.ylabel("Investor")

plt.show()
```

```python
yearly_funding = df.groupby("Year")["InvestmentAmount_USD"].sum()
yearly_funding
```

```python
growth = yearly_funding.pct_change() * 100
growth
```

```python
growth.plot(kind="line", marker="o")

plt.title("Year-over-Year Investment Growth")
plt.xlabel("Year")
plt.ylabel("Growth (%)")

plt.show()
```
