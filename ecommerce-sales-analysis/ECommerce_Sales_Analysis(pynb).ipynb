**EDA**

**Which product category earns the most revenue?**


```python
category_revenue = df.groupby('product_category')['revenue'].sum().sort_values(ascending = False)
print(category_revenue)
```

**Which region sells the most?**

```python
region_revenue = df.groupby('region')['revenue'].sum().sort_values(ascending=False)
print(region_revenue)
```

```python
payment_revenue = df.groupby('payment_method')['revenue'].sum().sort_values(ascending=False)
```

```python
#Imagine sorting students into grade buckets based on their score — 0-40 (Fail), 40-60 (C), 60-80 (B), 80-100 (A).
#pd.cut does exactly that, but for your discount numbers instead of scores.
```

```python
# why and when to use this step:
# If yes (many unique numbers) → you need to bucket it first with pd.cut, THEN groupby
# If no (already just a few categories) → skip bucketing, groupby directly
```

```python
df['discount_bucket'] = pd.cut(df['discount'], bins=[0,0.1,0.2,0.3,0.4], labels=['0-10%','10-20%','20-30%','30-40%'])
discount_revenue = df.groupby('discount_bucket')['revenue'].mean()
print(discount_revenue)
```

```python
# Discount given     →     Average revenue per order
# 0-10%  (small discount)  →  ₹1,157
# 10-20%                   →  ₹1,058
# 20-30%                   →  ₹944
# 30-40% (big discount)    →  ₹841
```

```python
df['delivery_bucket'] = pd.cut(df['delivery_days'], bins = [0, 3, 6, 9, 12], labels = ['1-3 days','4-6 days','7-9 days','10-12 days'])
delivery_ratings = df.groupby('delivery_bucket', observed=False)['customer_rating'].mean()
print(delivery_ratings)
```
